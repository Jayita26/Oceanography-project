```python
df['Date'] = pd.to_datetime(df[['Year', 'Month', 'Day']])
df.set_index('Date', inplace=True)
fig = plt.figure(figsize=(15, 10))
features = ['AirT', 'WIND', 'SST', 'SSS', 'SLD']
titles = ['Air Temperature', 'Wind Speed', 'Sea Surface Temperature', 'Sea Surface Salinity', 'Sonic Layer Depth']
for i, col in enumerate(features):
    plt.subplot(5,1,i+1)
    plt.plot(df.index, df[col])
    plt.ylabel(col)
    plt.xlabel('Date')
    plt.title(titles[i])
    
fig.tight_layout()

plt.show()
```
